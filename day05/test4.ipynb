```python
#终于进入可视化了
# 折线图plot
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='SimHei'# 设置字体
# 创建图片,设置大小,相当于是一个画布
plt.figure(figsize=(10,5))

# 要绘图的数据
month=['一月','二月','三月','四月']
sales=[400,490,789,901]

#在每个坐标上面展示数值,plt.text(x坐标, y坐标, 显示的文字)
for x,y in zip(month,sales):
    print(x,y)
    plt.text(x,y+1,y,ha='center',va='bottom')

#其中label显示图例信息,linestyle可以控制显示的折线样式
plt.plot(month,sales,label='产品B',color='orange',linestyle='--',linewidth=2)# 前一个x轴,后一个y轴
plt.title('2026年销售情况',color='red')
#显示图表,需要使用show函数

# 添加横轴标签与纵轴标签
plt.xlabel('月份',fontsize=10)
plt.ylabel('万元',fontsize=10)
# 添加图例
plt.legend(loc='upper left')# 位于左上角

#添加网格线,看得更加清楚,alpha修改网格透明度
plt.grid(True,alpha=0.1)

#设置Y轴范围
plt.ylim(0,1000)
plt.show()
```

```python
#柱状图的学习

import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='SimHei'# 设置字体
# 创建图片,设置大小,相当于是一个画布
plt.figure(figsize=(10,5))

subjects=['语文','数学','英语','化学']
scores=[90,100,67,78]
plt.bar(subjects,scores,label='小红成绩')


plt.title('小红的2026成绩分布',fontsize=10,color='red')
plt.xlabel('科目',fontsize=10)
plt.ylabel('分数',fontsize=10)
plt.legend(loc='upper left')
#添加网格线,看得更加清楚,alpha修改网格透明度
plt.grid(axis='y',alpha=0.5)

#在每个坐标上面展示数值,plt.text(x坐标, y坐标, 显示的文字)
for x,y in zip(subjects,scores):
    print(x,y)
    plt.text(x,y+1,y,ha='center',va='bottom')

plt.ylim(0,150)
# 自动化排版
plt.tight_layout()
plt.show()
```

```python
# 饼图的学习
#终于进入可视化了
# 折线图plot
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family']='SimHei'# 设置字体
# 创建图片,设置大小,相当于是一个画布
plt.figure(figsize=(10,5))

context=['睡觉','娱乐','学习','运动','其他']
percentage=[10,20,67,30,80]
explod=[0,0,0.1,0,0.1]# 设置突出块的位置,
#注意定义格式,不一样了,要展示的标签需要使用labels来指定,由于%有特殊含义,所以需要两个百分号进行特殊转义一下
# 使用wedgeprops将饼图改成环图
# plt.pie(percentage,labels=context,autopct='%.2f%%',wedgeprops={'width':0.6})
plt.pie(percentage,labels=context,autopct='%.2f%%',explode=explod,shadow=True)
plt.title('一天的时间分配',fontsize=10)
plt.show()
```

```python
# 散点图的学习,与折线图很像
#终于进入可视化了
# 散点图plot
import matplotlib.pyplot as plt
from matplotlib import rcParams
import numpy as np
rcParams['font.family']='SimHei'# 设置字体
# 创建图片,设置大小,相当于是一个画布
plt.figure(figsize=(10,5))

# 要绘图的数据
# month=['一月','二月','三月','四月']
# sales=[400,490,789,901]
# plt.scatter(month,sales)
x=[]
y=[]
for i in range(1000):
    tmp=np.random.uniform(0,10)
    x.append(tmp)# 如果不使用tmp记录的话,全部是随机的,根本就看不出来关系
    y.append(2*tmp+np.random.uniform(0,10))
plt.scatter(x,y,label='数据')


# #在每个坐标上面展示数值,plt.text(x坐标, y坐标, 显示的文字)
# for x,y in zip(x,y):
#     print(x,y)
#     plt.text(x,y+1,y,ha='center',va='bottom')

#其中label显示图例信息,linestyle可以控制显示的折线样式

plt.title('X与Y的关系',color='red')
#显示图表,需要使用show函数

# 添加横轴标签与纵轴标签
plt.xlabel('横坐标',fontsize=10)
plt.ylabel('纵坐标',fontsize=10)
# 添加图例
plt.legend(loc='upper left')# 位于左上角

#添加网格线,看得更加清楚,alpha修改网格透明度
plt.grid(True,alpha=0.1)

#设置Y轴范围
plt.ylim(0,30)
# 画一条直线,一个框内居然不是一个坐标,第一个框都是横坐标,第二个框都是纵坐标
plt.plot([0,10],[0,30])
plt.show()
```

```python
# 绘制多个图表使用subplot函数
month=['一月','二月','三月','四月']
sales=[400,490,789,901]
subjects=['语文','数学','英语','化学']
scores=[90,100,67,78]
context=['睡觉','娱乐','学习','运动','其他']
percentage=[10,20,67,30,80]
f1=plt.subplot(2,2,1)# 最后一个是图表的编号
f2=plt.subplot(2,2,2)# 最后一个是图表的编号
f3=plt.subplot(2,2,3)# 最后一个是图表的编号,前面个两个就是行数和列数,就是将一个画布切割成了几块
f1.plot(month,sales)
f2.bar(subjects,scores)
f3.barh(context,percentage)
f4=plt.subplot(2,2,4)
f4.pie(percentage,labels=context)

```
