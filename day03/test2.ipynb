```python
# 下面开始介绍pandas
# 先介绍第一个数据结构就是Series
import pandas as pd
import numpy as np
a=pd.Series([1,2,3,4])
print(a)
# 还可以使用自定义索引，就是左边的index,还可以取名字，就是描述这个数据是干什么用的
a=pd.Series([1,2,3,4],index=['A','B','C','D'],name='测试数据')
print(a)

# 通过字典创建一个serises,字典是使用大括号的
# 但是字典的key如果不加引号的话，就会是认作变量，最好还是加上引号
b=pd.Series({"name":"哈哈哈" ,"age":17,"address":"Unnkown","telephone":"88888888888"})
print(b)

#通过已有的series来创建自己的series
c=pd.Series(b,index=["name","address"])
print(c)
```

```python
# 接下来介绍一下series的属性，包括index，values，ndim，shape之类的
# 显式索引就是loc函数，可以根据索引定位数值,甚至支持切片（左闭右闭）
print(b.loc["name"])
print(b.loc["name":"address"])
# 隐式索引就是默认从0开始的index,同样支持切片
print(b.iloc[2])
# at函数，其实和loc一样，就是不支持切片
print(b.at["name"])
print(b.iat[0])
```

```python
# 访问Series里面的元素
# 直接使用标签访问即可
print(b["name"])
# 也可以使用bool索引,b好像不行，既有字符串又有数字
print(a[a>2])

a['e']=19
a['f']=90
# 使用head函数查看series的前五行数据,tail查看末尾五行数据，当然括号里面也可以写上数字
print(a)
print(a.head(6))
print(a.tail(3))
```

```python
# Series的常见方法
# describe函数，查看所有的描述性信息
a.describe()

#获取索引
print(a.index)# 使用属性
print(a.keys())# 使用方法

print(a)

print(a.isin([2,90]))# 查看列表中的元素是否在Series之中

# 打印众数函数mode
print(a.mode())
```

```python
# 按照索引排序以及values排序
a.sort_index()
# a.sort_values()
```

```python
# 创建一个包含10名学生的Series，成绩在50-100之间
import numpy as np
np.random.seed(20)
# 后面的那个写法是列表推导式，每次处理相当于使用了append函数
Scores=pd.Series(np.random.randint(50,101,10),index=['Student'+str(i) for i in range(1,11)])
print(Scores)

mean=Scores.mean()
Scores>mean# 这也行？后面再次使用bool索引
Scores[Scores>mean]
# Scores[Scores>mean].count()
```

```python
# 某个城市的气温
Tempature=pd.Series([29,31,28,37,32,40,29],index=['周一','周二','周三','周四','周五','周六','周日'],name=
'城市气温')
print(Tempature[Tempature>30])
Tempature.sort_values(ascending=False)

# 遍历Series的几种方式，如果使用单纯的for循环，遍历的是values，遍历index使用s.index
# 常见的方式：for a,b in s.items():其中a就可以取得index，b就可以取得values

#找出温差最大的两天，可以使用diff函数，获取后一个数字与其前一个数字的差值
a=Tempature.diff().abs()
print(a)
print(a.max(),a.min())
print(f"温度变化最大的两天：{a.sort_values(ascending=False).keys()[0:2].tolist()}")
```

```python
# 股票市值
date=pd.date_range('2026-09-4',periods=10)
Prices=pd.Series([102.3,104.5,104.8,103.9,105.9,103.4,105.6,104.7,104.3,105.6],index=date,name='股票市值')
print(Prices)
print(Prices['2026-09-05'])
# print(Prices[1])# 这种方式是错误的，需要使用iloc

for i in range(9,-1,-1):
    if i==0:
        Prices.iloc[i]=0
    else:
        Prices.iloc[i]=Prices.iloc[i]/Prices.iloc[i-1]-1
print(Prices)
# 找出收益最高与最低的日子,根据value获取key，使用bool索引,也可以使用idxmax，我真是服了
print(Prices[Prices==Prices.max()].index)
print(Prices.idxmax())

# Prices.pct_change()
```
