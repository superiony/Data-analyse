```python
#下面开始介绍DataFrame，就是pandas里面的二位数据结构
import pandas as pd
# dataframe要求每一列可以是不同的数据类型，但是同一列的话必须是同一种数据类型
# 相当于是每一列都是一个Series
s1=pd.Series([1,2,4,6,7])
s2=pd.Series([9,10,65,78,90])
df=pd.DataFrame({"第一列":s1,"第二列":s2},index=[2,3,4,0,1],columns=['第二列','第一列'])# 需要按照字典的方式创建Dataframe
df
#注意创建DataFrame的时候，index默认从0开始，index填入的数值就是要展示的行数以及顺序
# 同理，columns要填入的数值就是要展示的列的顺序以及展示那些列
```

```python
# 关于DataFrame的属性操作
print(f"data的行索引：{df.index}")
print(f"data的列索引：{df.columns}")
df.T
print(df.T.index,df.T.columns)
```

```python
# 关于DataFrame的索引操作，loc iloc at，其中loc是显示操作，iloc就是隐式操作
print(df.loc[4])
print(df.iloc[3])

df
#获取某一列操作
print(df.loc[:,'第二列'])
print(df.iloc[:,0])# 使用隐式位置来索引第一列的数据



#使用at与iat来获取某一个具体的数值
print(df.at[4,'第二列'])
print(df.iat[2,0])
```

```python
#也可以使用index直接获取某一列的数据
print(df['第二列'])
print(df.第二列)# 居然还可以这样子


# 获取多列数据，就需要使用两层括号，返回的是一个dataframe
print(df[['第一列','第二列']])
df[['第一列','第二列']]

# 使用head与tail函数也是可以看到这个部分数据的
# 同时也是可以使用bool索引的,中括号写法也可以，不过要加上这个字符串
df[df.第二列>40]
```
